```julia
function vdw(x, T)
    y =8 .* T ./ (3 *x .- 1) .- 3 ./ (x.*x) .-0.9
    return y
end

v=0.4:0.01:50
p = vdw(v, 4/5)

f = Figure()
ym = max(abs(minimum(p)), 2)


ax = Axis(f[1, 1], xlabel = "x label", ylabel = "y label", xscale = log10, 
    title = "Title")

lines!(ax, v, p)
ylims!(ax, -1.1*abs(ym), 1.5*abs(ym))

f
```
