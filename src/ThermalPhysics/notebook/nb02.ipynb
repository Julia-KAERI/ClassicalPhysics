```julia
T = [100, 200, 300, 400, 500, 600];
B = [-160, -35, -4.2, 9.0, 16.9, 21.3].* 1.0e-6
```

```julia
T
```

```julia
bb(t, a) = a[1] .- a[2]./t  
```

```julia
p0= [0.5, 0.5]
fit = curve_fit(bb, T, B, p0)
```

```julia
f = Figure()
ax = Axis(f[1, 1], xlabel = "T [K]", ylabel = "B(T)")

scatter!(ax, T, B, color=:red)
t=90:1:600
lines!(ax, t, bb(t, fit.param))
f
save("schroeder_prob_1_17.png", f)
```
